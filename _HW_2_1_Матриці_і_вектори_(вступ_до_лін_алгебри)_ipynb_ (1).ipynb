### Завдання 1: Створення матриць та векторів
1. Створіть `numpy` вектор `vector` розмірністю 10 з випадковими дійсними значеннями в діапазоні від 0 до 1. Можна скористатись двома альтернативними варіантами:
  1. Генерація випадкових чисел з модулем [`random.uniform`](https://docs.python.org/3/library/random.html).
  2. Створення вектора з допомогою метода `numpy` [np.random.rand](https://numpy.org/doc/stable/reference/random/generated/numpy.random.rand.html)

2. Створіть матрицю `matrix` типу `numpy.array` розмірності 3x3 з наступними значеннями

  $$
  matrix = \begin{bmatrix}
  11 & 14 & 17 \\
  20 & 23 & 26 \\
  29 & 32 & 35
  \end{bmatrix}
  $$

  будь-яким відомими Вам способом.

Виведіть на екран вектор і матрицю, а також їх розміри.


```python
import numpy as np
import random

# Варіант 1: створення вектора з random.uniform
vector1 = np.array([random.uniform(0, 1) for _ in range(10)])

# Варіант 2: створення вектора з np.random.rand
vector2 = np.random.rand(10)

# Створення матриці 3x3
matrix = np.array([
    [11, 20, 29],
    [14, 23, 32],
    [17, 26, 35]
])

# Вивід результатів
print("Вектор (random.uniform):\n", vector1)
print("Розмір вектора:", vector1.shape)

print("\nВектор (np.random.rand):\n", vector2)
print("Розмір вектора:", vector2.shape)

print("\nМатриця:\n", matrix)
print("Розмір матриці:", matrix.shape)
```

Пояснення:

random.uniform(0, 1) — генерує випадкове число від 0 до 1.

np.random.rand(10) — створює вектор з 10 випадкових значень від 0 до 1.

np.array([...]) — створює матрицю з заданими значеннями.

.shape — дозволяє дізнатися розмірність масиву (вектора або матриці).


### Завдання 2: Операції над векторами та матрицями
Виконайте кожну з зазначених операцій над зазначеними оʼєктами і виведіть результат на екран:
1. Додайте до кожного значення вектора `vector` число 10 і округліть кожне значення до 2 знаків після коми без використання циклів.
2. Помножте кожен елемент матриці з першого завдання на 2 без використання циклів.
3. Створіть новий вектор, який містить квадрат кожного `vector` без циклів.


```python
import numpy as np
import random

# Створення вектора та матриці (як у Завданні 1)
vector = np.random.rand(10)
matrix = np.array([
    [11, 20, 29],
    [14, 23, 32],
    [17, 26, 35]
])

# 1. Додаємо 10 до кожного елементу вектора та округлюємо до 2 знаків після коми
vector_plus_10 = np.round(vector + 10, 2)

# 2. Множимо кожен елемент матриці на 2
matrix_times_2 = matrix * 2

# 3. Створюємо новий вектор – квадрат кожного елемента початкового вектора
vector_squared = vector ** 2

# Вивід результатів
print("Оригінальний вектор:\n", vector)
print("\nВектор + 10 (округлений до 2 знаків):\n", vector_plus_10)

print("\nОригінальна матриця:\n", matrix)
print("\nМатриця, помножена на 2:\n", matrix_times_2)

print("\nВектор квадратів кожного елементу:\n", vector_squared)
```

Пояснення:

vector + 10 — додає число 10 до кожного елемента.

np.round(..., 2) — округлює всі значення до 2 знаків після коми.

matrix * 2 — множить всі елементи матриці на 2.

vector ** 2 — підносить кожен елемент вектора до квадрату.

Це все реалізовано через векторизовані операції, без використання циклів, що є перевагою NumPy.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

1. Множення матриць A (3x2) та B (2x3)

```python
import numpy as np

# Задані матриці
A = np.array([
    [0.18, 0.53],
    [0.75, 0.47],
    [0.35, 0.21]
])  # A: 3x2

B = np.array([
    [0.17, 0.58, 0.75],
    [0.38, 0.11, 0.15]
])  # B: 2x3

C = np.array([[0], [1]])  # 2x1
D = np.array([[0, 1], [1, 0]])  # 2x2

# Правильне множення: (2x3) @ (3x2) = (2x2)
result_matrix = B @ A

# Далі множення з C і D
result_C = result_matrix @ C  # (2x2) @ (2x1) = (2x1)
result_D = result_matrix @ D  # (2x2) @ (2x2) = (2x2)

# Вивід результатів
print("Матриця результату B @ A:\n", result_matrix)
print("\nresult_matrix @ C:\n", result_C)
print("\nresult_matrix @ D:\n", result_D)
```


### Завдання 4: Транспонування матриць
1. Транспонуйте `result_matrix`.
2. Зробіть транспонування вектора `vector` і виведіть форму нової структури. Чи бачите ви зміну?
3. Використайте операцію `numpy.expand_dims` над вектором `vector` з аргументом axis=1. Запишіть отриманий результат в змінну `column_vector` і виведіть на екран. Порівняйте з `vector`.
4. Зробіть транспонування `column_vector`, виведіть результат на екран разом з розмірностями отриманої структури.




```python
import numpy as np

# Створимо для прикладу vector
vector = np.random.rand(10)  # (10,)

# Припустимо, що result_matrix отриманий раніше:
result_matrix = np.array([[1.2, 0.8],
                          [0.9, 0.6]])  # Приклад (2×2)

# 1. Транспонування матриці
result_matrix_T = result_matrix.T
print("Транспонована result_matrix:\n", result_matrix_T)

# 2. Транспонування вектора
vector_T = vector.T
print("\nТранспонований vector:\n", vector_T)
print("Форма vector:", vector.shape)
print("Форма vector_T:", vector_T.shape)

# Пояснення: у NumPy одновимірний вектор не змінює форму при транспонуванні

# 3. Перетворення в стовпчиковий вектор
column_vector = np.expand_dims(vector, axis=1)
print("\nСтовпчиковий вектор column_vector:\n", column_vector)
print("Форма column_vector:", column_vector.shape)

# 4. Транспонування column_vector
column_vector_T = column_vector.T
print("\nТранспонований column_vector:\n", column_vector_T)
print("Форма column_vector_T:", column_vector_T.shape)
```

🔍 Висновок:

Одновимірні масиви в NumPy не змінюють форму при .T.

Щоб отримати “вектор-рядок” або “вектор-стовпчик” — треба явно змінити розмірність.

expand_dims(vector, axis=1) або reshape(-1, 1) — це правильний спосіб зробити вектор-стовпчик.
